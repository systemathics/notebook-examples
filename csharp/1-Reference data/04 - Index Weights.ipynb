# [C#] - Index Weights

In this tutorial, we'll provide a comprehensive guide on retrieving both the index **components** and their respective **weights** for a specified **index**. 

## Step 1 : Setup and import packages

### Step 1: Install packages

```
#i "nuget: file:///home/jovyan/.nuget/packages/"
#i "nuget: https://api.nuget.org/v3/index.json"
#r "nuget: Systemathics.Apis, 2.39.*"
#r "nuget: XPlot.Plotly.Interactive, 4.0.7"
```

```
// Import required packages
// Systemathics packages
using Systemathics.Apis.Helpers;
using Systemathics.Apis.Type.Shared.V1;
using Systemathics.Apis.Services.StaticData.V1;

// Essential packages 
using Google.Protobuf.WellKnownTypes;
using Google.Type;
using Grpc.Net.Client;
using Grpc.Core;

// Plot graphs
using XPlot.Plotly;
using XPlot.Plotly.Interactive;
```

## Step 2: Retrieve a valid token

```
// Get authentication token
var headers = TokenHelpers.GetTokenAsMetaData();

// Create communication channel
var channel = ChannelHelpers.GetChannel(); 
```

## Step 3: Call Indices Service

```
// Instantiate the service
var service = new StaticDataService.StaticDataServiceClient(channel);
```

## Step 4: Create and process request
In this sample, we presuppose familiarity with the ticker symbol and exchange of the index under investigation. Let's exemplify this with the renowned **CAC 40 Index**.

*For more details on how to list available indices, please refer to the "Get Started" tutorial.*

```
// Generate the indice components request
var request = new StaticDataRequest 
{
    AssetType = AssetType.Index,
    Exchange = "XPAR", 
    Ticker = "PX1"
};
```

```
// Process the indice components response
var reply = service.StaticData(request, headers);
```

```
// Display the results
display(reply.Index.Assets.FirstOrDefault().Components)
```

## 5 Visualize Index data

The following code snippet displays index components and weights. <br>*`XPlot.Plotly`* is used as per open source display package. 

```
var data = reply.Index.Assets.FirstOrDefault().Components.Select(elt => new Tuple<String, double>(elt.Key, elt.Value));
```

```
var chart = Chart.Pie(data);
chart.WithTitle($"Components and weights");
chart.Width = 1500;
chart.Height = 800;
display(chart);
```
