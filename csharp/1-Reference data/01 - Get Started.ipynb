## Step 1 : Setup and import packages
Before proceeding, make sure to set up and import the appropriate packages.

```
#i "nuget: file:///home/jovyan/.nuget/packages/"
#i "nuget: https://api.nuget.org/v3/index.json"
#r "nuget: Systemathics.Apis, 2.39.*"
```

```
// Import required packages
// Systemathics packages
using Systemathics.Apis.Type.Shared.V1;
using Systemathics.Apis.Helpers;
using Systemathics.Apis.Services.StaticData.V1;

// Essential packages 
using Google.Protobuf.WellKnownTypes;
using Google.Type;
using Grpc.Net.Client;
using Grpc.Core;
```

## Step 2: Retrieve a valid token
To proceed, you need to retrieve a valid authentication *token*. 
<br><br>The following code snippets demonstrate how to obtain an authentication *token* and prepare the API request. This involves opening the communication *channel* and adding the token to the request header:

```
// Get authentication token
var headers = TokenHelpers.GetTokenAsMetaData();

// Create communication channel
var channel = ChannelHelpers.GetChannel(); 
```

## Step 3: Call Static Data Service
With the setup complete, we're now ready to initiate the service.

```
// Instantiate the service
var service = new StaticDataService.StaticDataServiceClient(channel);
```

## Step 4: Create and process request

### 4.1 Request Data by Asset Type
When querying data by asset type, you're specifying the broad category of financial instruments you're interested in. For example, you can request data for equities, futures, options, bonds, cryptocurrencies, and more. By setting the *AssetType* field in the request object, you filter the returned data to include only instruments of that specific type.

For equities for example, you need to specify the AssetType as <i>`AssetType.Equity`</i> in the static data request.

To retrieve all available asset types, please refer to the [Ganymede documentation](https://ganymede.cloud/docs/#systemathics/apis/services/static_data/v1/assettype). 

The following snippet displays all the available values of the AssetType enum.

```
// Display all AssetType Enum Values
Console.WriteLine("All AssetType Enum Values:");
foreach (AssetType assetType in System.Enum.GetValues(typeof(AssetType)))
{
    Console.WriteLine(assetType);
}
```

In the following snippet, we demonstrate how to request all equities as an example. By default, the maximum records returned is set to 500. 

To handle larger datasets, *pagination* is necessary. Please refer to the sample pagination implementation for guidance.

```
// Generate request
var request = new StaticDataRequest
{
  AssetType = AssetType.Equity //<-- Insert the asset type here
};
```

```
// handle the response
var reply = service.StaticData(request, headers);
```

```
// Display the response : counts
display($"Total count = {reply.Equity.TotalCount}");
display($"Found count = {reply.Equity.Assets.Count}");
```

```
// Display the response : the first item from assets
display(reply.Equity.Assets.FirstOrDefault())
```

### 4.2 Request Data by Name
Searching for financial instruments by name allows you to retrieve data for a specific company or asset based on its name. This can be particularly useful when you want to fetch information about a well-known company or a specific asset without knowing its ticker symbol.
<br><br>
Apple Inc. is a well-known technology company, and its stock is widely traded on various stock exchanges. In this example, we'll demonstrate how to request static data for Apple Inc. using its name "Apple".

```
var request = new StaticDataRequest
{
  AssetType = AssetType.Equity,
  Name = "Apple" //<-- Include the asset name here
};
```

```
var reply = service.StaticData(request, headers);
```

```
display(reply.Equity.Assets.FirstOrDefault().Identifier)
```

### 4.3 Request Data by Exchange
When querying data by **Exchange**, you're focusing on instruments traded on a particular exchange. Exchanges are platforms where buyers and sellers come together to trade financial instruments. By setting the Exchange field in the request object, you filter the returned data to include only instruments traded on the specified exchange.
<br><br>
In the snippet below, we also demonstrate how to specify the desired number of records using the *Count* property and the starting index using the *Start* property in the *request*.

```
var request = new StaticDataRequest
{
  AssetType = AssetType.Equity,
  Exchange = "XNAS", //<-- Include the exchange mic code
  Count = 10, // Select the first n elements
  Start = 5 // Starting from element index at i
};
```

```
var reply = service.StaticData(request, headers);
```

```
// Display the response : the first item identifier from assets
display(reply.Equity.Assets.FirstOrDefault().Identifier)
```

### 4.4 Request Data by Ticker
Querying data by ticker allows you to retrieve information about a specific financial instrument identified by its ticker symbol. Ticker symbols uniquely identify assets on various financial markets. For example, "AAPL" represents Apple Inc. stock on most exchanges. By setting the Ticker field in the request object, you retrieve static data for the instrument with the specified ticker symbol.

```
var request = new StaticDataRequest
{
  AssetType = AssetType.Equity,
  Ticker = "AAPL"
};
```

```
var reply = service.StaticData(request, headers);
```

```
display(reply.Equity.Assets.FirstOrDefault().Identifier)
```

### 4.5 Request Data by ISIN
**ISINs** are unique identifiers assigned to securities to facilitate the clearance and settlement process. They provide a standardized format for identifying financial instruments globally. In this example, we'll demonstrate how to request static data for a specific instrument using its ISIN.
<br><br>
The purpose of this example is to showcase how you can retrieve information about a financial instrument using its ISIN. This can be beneficial when you have the ISIN of a security and need to fetch its static data, such as its ticker symbol, name, and asset type.

```
var request = new StaticDataRequest
{
  AssetType = AssetType.Equity,
  Isin = "FR0000131104" //<-- Insert the ISIN code here
};
```

```
var reply = service.StaticData(request, headers);
```

```
display(reply.Equity.Assets.FirstOrDefault().Identifier)
```

### 4.6  Request Data Using Multiple Filters
Sometimes, you may need to narrow down your search by using multiple filters to retrieve specific data. In this example, we'll showcase how to request static data for an instrument using both its ticker symbol and the exchange it is traded on.
<br><br>
The purpose of this example is to demonstrate how you can combine multiple filters to refine your data query and retrieve more precise results. By specifying both the *Ticker* symbol and *Exchange*, you can ensure that the data returned matches the exact instrument you're interested in.

```
// The service also enables to query with multiple filters as follows
var request = new StaticDataRequest
{
  AssetType = AssetType.Equity,
  Ticker = "AAPL", //<-- The ticker code
  Exchange = "XNAS" //<-- The exchange mic code
};
```

```
var reply = service.StaticData(request, headers);
```

```
display(reply.Equity.Assets.FirstOrDefault().Identifier)
```
